# 两数之和与三数之和

两数之和用哈希表查找补数，平均 O(n) 时间。三数之和先排序，固定一个数后，用双指针在后缀中找其相反数，时间 O(n²)。跳过重复值来保证三元组唯一。

```
def two_sum(nums, target):
    seen = {}
    for index, value in enumerate(nums):
        if target - value in seen:
            return seen[target - value], index
        seen[value] = index
    return None


def three_sum(nums):
    ordered = sorted(nums)
    answer = []
    for index, first in enumerate(ordered[:-2]):
        if first > 0:
            break
        if index and first == ordered[index - 1]:
            continue
        left, right = index + 1, len(ordered) - 1
        while left < right:
            total = first + ordered[left] + ordered[right]
            if total < 0:
                left += 1
            elif total > 0:
                right -= 1
            else:
                answer.append([first, ordered[left], ordered[right]])
                left += 1
                right -= 1
                while left < right and ordered[left] == ordered[left - 1]:
                    left += 1
    return answer


assert two_sum([3, 3], 6) == (0, 1)
assert three_sum([-1, 0, 1, 2, -1, -4]) == [[-1, -1, 2], [-1, 0, 1]]
assert three_sum([0, 0, 0, 0]) == [[0, 0, 0]]
```
