# 前缀和

令 `prefix[i]` 为前 `i` 个元素之和。构建需要 O(n) 时间和空间，半开区间 `[left, right)` 的和用 `prefix[right] - prefix[left]` 在 O(1) 内求出。前缀和可能因负数而下降。

统计和为 `target` 的连续子数组时，当前前缀为 `s`，每个更早出现的 `s - target` 都对应一个答案。先查再计数，以排除空子数组。

```
from collections import Counter
from itertools import accumulate


def range_sum(prefix, left, right):
    return prefix[right] - prefix[left]


def count_subarrays(nums, target):
    seen = Counter({0: 1})
    prefix = answer = 0
    for value in nums:
        prefix += value
        answer += seen[prefix - target]
        seen[prefix] += 1
    return answer


prefix = list(accumulate([3, -2, 5, 1], initial=0))
assert range_sum(prefix, 1, 3) == 3
assert range_sum(prefix, 0, 0) == 0
assert count_subarrays([1, -1, 1], 1) == 3
assert count_subarrays([0, 0], 0) == 3
```
