# 双指针

在升序数组中找两数之和时，从两端向内收缩。当前和太小，左端与任何尚未排除的右端组合都太小，因此左移除。同理，和太大就移除右端。只遍历一次，额外空间为 O(1)。

快慢指针适合原地整理数组。对有序数组去重时，快指针逐个读，慢指针记住下一个写入位置。

```
def two_sum_sorted(nums, target):
    left, right = 0, len(nums) - 1
    while left < right:
        total = nums[left] + nums[right]
        if total == target:
            return left, right
        if total < target:
            left += 1
        else:
            right -= 1
    return None


def deduplicate_sorted(nums):
    write = 0
    for value in nums:
        if write == 0 or nums[write - 1] != value:
            nums[write] = value
            write += 1
    return write


assert two_sum_sorted([1, 2, 4, 9], 6) == (1, 2)
assert two_sum_sorted([1], 2) is None
values = [1, 1, 2, 2, 3]
assert values[:deduplicate_sorted(values)] == [1, 2, 3]
assert deduplicate_sorted([]) == 0
```
