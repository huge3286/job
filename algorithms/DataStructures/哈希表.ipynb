# 哈希表

`dict` 存键和值，`set` 存唯一的键。`Counter` 是计数用的字典。插入和查询平均 O(1)，键必须可哈希。

判断数组中是否有距离不超过 `k` 的重复元素时，只需保留每个值上次出现的位置。

```
from collections import Counter


def contains_nearby_duplicate(nums, k):
    last = {}
    for index, value in enumerate(nums):
        if value in last and index - last[value] <= k:
            return True
        last[value] = index
    return False


assert Counter("abca") == {"a": 2, "b": 1, "c": 1}
assert contains_nearby_duplicate([1, 2, 3, 1], 3)
assert not contains_nearby_duplicate([1, 2, 3, 1], 2)
assert not contains_nearby_duplicate([], 0)
```
