# 堆

堆是完全二叉树。小根堆的每个节点不大于其子节点，因此堆顶是最小值。`heapify` 原地构建堆的时间为 O(n)，插入和弹出堆顶为 O(log n)。

只维护最大的 k 个数时，可以用大小不超过 k 的小根堆。更大的候选替换堆顶，总时间 O(n log k)。

```
from heapq import heapify, heappop, heappush, heapreplace


def largest_k(nums, k):
    if k <= 0:
        return []
    heap = []
    for value in nums:
        if len(heap) < k:
            heappush(heap, value)
        elif value > heap[0]:
            heapreplace(heap, value)
    return sorted(heap, reverse=True)


heap = [8, 3, 5, 1]
heapify(heap)
assert heappop(heap) == 1
assert largest_k([14, 20, 5, 28, 28], 3) == [28, 28, 20]
assert largest_k([1], 0) == []
```
