# 栈

后进先出。Python 的 `list.append` 和尾部 `pop` 可实现栈。括号匹配时保存待闭合的括号。

单调栈还可用于寻找右侧第一个更大元素。栈内保留等待答案的下标，每个下标至多入栈、出栈各一次，时间 O(n)。

```
def valid_brackets(text):
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []
    for char in text:
        if char in pairs.values():
            stack.append(char)
        elif char not in pairs or not stack or stack.pop() != pairs[char]:
            return False
    return not stack


def next_greater(nums):
    answer = [-1] * len(nums)
    stack = []
    for index, value in enumerate(nums):
        while stack and nums[stack[-1]] < value:
            answer[stack.pop()] = value
        stack.append(index)
    return answer


assert valid_brackets("([]{})")
assert not valid_brackets("([)]")
assert next_greater([2, 1, 2, 4, 3]) == [4, 2, 4, -1, -1]
```
