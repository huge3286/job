## 链表

```python
from typing import Optional
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
class Solution:
    def doubleIt(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if head is None:
            return None
        if head.val > 4:
            head = ListNode(0, head)
        cur = head
        while cur:
            cur.val = cur.val * 2 % 10
            if cur.next and cur.next.val > 4:
                cur.val += 1
            cur = cur.next
        return head
```

```python
result = Solution().doubleIt(ListNode(9, ListNode(5)))
assert (result.val, result.next.val, result.next.next.val) == (1, 9, 0)
assert Solution().doubleIt(None) is None
```
