# 队列

先进先出。用 `collections.deque` 从队尾追加、队首弹出，两端操作为 O(1)。用队列做无权图 BFS 时，节点在入队时标记，首次发现它的距离就是最短距离。

```
from collections import deque


def bfs_distances(graph, start):
    distances = {start: 0}
    queue = deque([start])
    while queue:
        node = queue.popleft()
        for neighbor in graph.get(node, []):
            if neighbor not in distances:
                distances[neighbor] = distances[node] + 1
                queue.append(neighbor)
    return distances


assert bfs_distances({0: [1, 2], 1: [2], 2: [0], 3: []}, 0) == {0: 0, 1: 1, 2: 1}
assert bfs_distances({}, 0) == {0: 0}
```
