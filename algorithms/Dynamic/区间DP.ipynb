# 区间 DP

子问题用左右端点表示，按长度从短到长计算。以合并相邻石堆为例，最后一次合并会把区间分成两段，总代价为左右最优代价加区间重量和。

枚举 O(n²) 个区间及每个区间的 O(n) 个切分点，时间 O(n³)，空间 O(n²)。

```
from itertools import accumulate


def min_merge_cost(stones):
    n = len(stones)
    if n < 2:
        return 0
    prefix = list(accumulate(stones, initial=0))
    dp = [[0] * n for _ in stones]
    for length in range(2, n + 1):
        for left in range(n - length + 1):
            right = left + length - 1
            dp[left][right] = prefix[right + 1] - prefix[left] + min(
                dp[left][split] + dp[split + 1][right]
                for split in range(left, right)
            )
    return dp[0][-1]


assert min_merge_cost([3, 2, 4, 1]) == 20
assert min_merge_cost([5]) == 0
assert min_merge_cost([]) == 0
```
