# 树形 DP

先计算子节点，再合并到父节点。以二叉树最大独立集权重为例，每个节点保留「选当前节点」和「跳过当前节点」两种状态。选当前节点时只能跳过两个子节点。跳过当前节点时，每棵子树独立选更优状态。时间 O(n)，递归栈 O(树高)。

```
def maximum_independent_weight(root):
    # 每个节点为 (权重, 左子树, 右子树)。
    def visit(node):
        if node is None:
            return 0, 0
        weight, left, right = node
        left_take, left_skip = visit(left)
        right_take, right_skip = visit(right)
        return weight + left_skip + right_skip, max(left_take, left_skip) + max(right_take, right_skip)

    return max(visit(root))


tree = (3, (2, None, (3, None, None)), (3, None, (1, None, None)))
assert maximum_independent_weight(tree) == 7
assert maximum_independent_weight(None) == 0
```
