# 线性 DP

用下标描述已处理前缀。打家劫舍问题中，走到当前房屋时，可跳过它并沿用前一状态，或选它并加上前两步的最优值。只需要保留最近两个状态，时间 O(n)，空间 O(1)。

```
def rob(values):
    two_back = one_back = 0
    for value in values:
        two_back, one_back = one_back, max(one_back, two_back + value)
    return one_back


assert rob([2, 7, 9, 3, 1]) == 12
assert rob([1, 2, 3, 1]) == 4
assert rob([]) == 0
```
