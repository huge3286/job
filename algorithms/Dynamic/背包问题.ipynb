# 背包问题

0–1 背包每件物品至多选一次，容量倒序更新，以便读取加入当前物品前的状态。完全背包允许重复选物品，容量正序更新。下列一维写法的时间为 O(n × capacity)，空间为 O(capacity)。

```
def knapsack_01(items, capacity):
    dp = [0] * (capacity + 1)
    for weight, value in items:
        for remaining in range(capacity, weight - 1, -1):
            dp[remaining] = max(dp[remaining], dp[remaining - weight] + value)
    return dp[capacity]


def coin_change(coins, amount):
    unreachable = amount + 1
    dp = [0] + [unreachable] * amount
    for coin in coins:
        for total in range(coin, amount + 1):
            dp[total] = min(dp[total], dp[total - coin] + 1)
    return dp[amount] if dp[amount] < unreachable else -1


assert knapsack_01([(2, 3), (3, 4), (4, 8)], 5) == 8
assert knapsack_01([(2, 3)], 4) == 3
assert coin_change([1, 2, 5], 11) == 3
assert coin_change([2], 3) == -1
```

## 题目变体

目标和使用「当前和 → 方案数」记录每轮加减之后的结果。分割等和子集是 0–1 背包的可行性版本。零钱兑换 II 的外层遍历面额，避免把同一组合的不同排列重复计数。

```
from collections import Counter


def target_sum_ways(nums, target):
    counts = Counter({0: 1})
    for value in nums:
        updated = Counter()
        for total, count in counts.items():
            updated[total + value] += count
            updated[total - value] += count
        counts = updated
    return counts[target]


def can_partition(nums):
    total = sum(nums)
    if total % 2:
        return False
    target = total // 2
    return knapsack_01([(value, value) for value in nums], target) == target


def count_change(coins, amount):
    ways = [1] + [0] * amount
    for coin in coins:
        for total in range(coin, amount + 1):
            ways[total] += ways[total - coin]
    return ways[amount]


assert target_sum_ways([1, 1, 1, 1, 1], 3) == 5
assert target_sum_ways([0], 0) == 2
assert can_partition([1, 5, 11, 5])
assert not can_partition([1, 2, 3, 5])
assert count_change([1, 2, 5], 5) == 4
```
