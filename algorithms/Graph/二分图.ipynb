## 二分图 Bipartite Graph  
什么是二分图 每条边的两个端点分属两个互不相交的顶点集合 A 和 B  


### 二分图的判断  
染色法 相连的两个节点必须是不同颜色的
1. `dfs`
2. `bfs`  

两个方法大同小异 需要注意的是
1. `[]`视为二分图
2. 同一个集合之间可以不相连

```python
# dfs判断二分图
class Solution:
    def isBipartite(self, graph) -> bool:
        n = len(graph)
        colors = [0 for _ in range(n)]

        def dfs(i, color):
            colors[i] = color
            for j in graph[i]:
                if colors[j] == colors[i]:
                    return False
                if colors[j] == 0 and not dfs(j, -color):
                    return False
            return True

        for i in range(n):
            # 因为存在不连通的情况所有要找到所有没染色的节点
            # []空节点认为是染色了的
            if colors[i] == 0 and not dfs(i, 1):
                return False
        return True
```

## 二分图最大匹配

每个右侧点最多匹配一次。DFS 尝试寻找增广路，必要时为当前占用该右侧点的左侧点另找匹配。带权变体见 [状态压缩 DP](../Dynamic/状压DP.ipynb)。

```python
def maximum_matching(graph):
    # graph[left] 给出该左侧点能连接的右侧点。
    matched = {}

    def augment(left, visited):
        for right in graph[left]:
            if right in visited:
                continue
            visited.add(right)
            if right not in matched or augment(matched[right], visited):
                matched[right] = left
                return True
        return False

    for left in range(len(graph)):
        augment(left, set())
    return matched


assert Solution().isBipartite([[1], [0], []])
assert not Solution().isBipartite([[1, 2], [0, 2], [0, 1]])
assert maximum_matching([[0, 1], [0]]) == {0: 1, 1: 0}
```
