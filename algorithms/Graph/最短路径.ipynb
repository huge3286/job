# Dijkstra 最短路径

当边权非负时，堆顶弹出的最小有效距离已确定。更新邻居时把新的候选推入堆，弹出时丢弃过时记录。适用于有向图或无向图，邻接表中未列出的节点视为没有出边。

字典只返回从起点可达的节点。时间 O((V + E) log(V + E))，包括图在内的空间 O(V + E)。完整路径示例见 [Python](examples/dijkstra.py) 和 [C++](examples/dijkstra.cc)。

```
from heapq import heappop, heappush
from itertools import count
from math import inf


def dijkstra(graph, start):
    distance = {start: 0}
    sequence = count()
    queue = [(0, next(sequence), start)]
    while queue:
        current, _, node = heappop(queue)
        if current != distance[node]:
            continue
        for neighbor, weight in graph.get(node, ()):
            if weight < 0:
                raise ValueError("Dijkstra 要求边权非负")
            candidate = current + weight
            if candidate < distance.get(neighbor, inf):
                distance[neighbor] = candidate
                heappush(queue, (candidate, next(sequence), neighbor))
    return distance


graph = {"A": [("B", 10), ("C", 2)], "C": [("B", 3)], "D": []}
assert dijkstra(graph, "A") == {"A": 0, "C": 2, "B": 5}
assert dijkstra(graph, "D") == {"D": 0}
```
