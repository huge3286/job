# 买卖股票的最佳时机

只允许一次交易时，假设今天卖出，则买入价格是此前最低价。维护最低价和全局最大利润即可，时间 O(n)，空间 O(1)。

允许多次交易且无手续费时，每一段上涨都可以贡献利润，累加所有相邻的正价差。

```
from math import inf


def max_profit_once(prices):
    lowest = inf
    best = 0
    for price in prices:
        lowest = min(lowest, price)
        best = max(best, price - lowest)
    return best


def max_profit_unlimited(prices):
    return sum(max(0, current - previous) for previous, current in zip(prices, prices[1:]))


prices = [7, 1, 5, 3, 6, 4]
assert max_profit_once(prices) == 5
assert max_profit_unlimited(prices) == 7
assert max_profit_once([3, 2, 1]) == 0
assert max_profit_once([]) == 0
```
