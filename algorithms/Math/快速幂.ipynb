# 快速幂

把非负整数指数拆成二进制位。每右移一位时将底数平方，如果当前位为 1 就把当前底数乘入结果。可在每一步取模，时间 O(log n)。质数模 p 下，若 b 不被 p 整除，则 b 的逆元是 b^(p−2) mod p。

```
def quick_pow(base, exponent, modulus):
    if exponent < 0 or modulus <= 0:
        raise ValueError("指数须非负，模数须为正")
    result = 1 % modulus
    base %= modulus
    while exponent:
        if exponent & 1:
            result = result * base % modulus
        base = base * base % modulus
        exponent >>= 1
    return result


assert quick_pow(7, 123, 13) == pow(7, 123, 13)
assert quick_pow(4, 0, 1) == 0
assert 7 * quick_pow(7, 11, 13) % 13 == 1
```
